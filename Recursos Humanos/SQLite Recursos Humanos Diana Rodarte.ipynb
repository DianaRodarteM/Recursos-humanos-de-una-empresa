```python
import sqlite3
import pandas as pd
import os


# CSV

csv_path = "recursos_humanos.csv"

df = pd.read_csv(csv_path)

df.head()
```

```python
# Base de datos RH.db

db_path = "RH.db"

conn = sqlite3.connect(db_path)


# CSV a la tabla "Detalle"

df.to_sql("Detalle", conn, if_exists="replace", index=False)
```

```python
# Se despliegan las columnas sales, salary y satisfaction en orden descendente de acuerdo al campo “satisfaction_level"

consulta1 = """
SELECT sales, salary, satisfaction_level
FROM Detalle
ORDER BY satisfaction_level DESC;
"""

q1 = pd.read_sql_query(consulta1, conn)
q1
```

```python
# Promedio obtenido de la última evaluación para cada departamento

consulta2 = """
SELECT sales, AVG(last_evaluation) AS promedio_ultima_evaluacion
FROM Detalle
GROUP BY sales;
"""

q2 = pd.read_sql_query(consulta2, conn)
q2
```

```python
# Departamentos para los cuales el promedio de accidentes de trabajo sea mayor a 0.15

consulta3 = """
SELECT sales, AVG(Work_accident) AS promedio_accidentes
FROM Detalle
GROUP BY sales
HAVING AVG(Work_accident) > 0.15
"""
pd.read_sql(consulta3, conn)
```
